```python
import pandas as pd 
#pandas module for cleaning dataset
df = pd.read_csv("C:\SalesProject.csv",encoding = "latin-1")
df.info()
# just to check 
```

```python
df.info()
# info() - all the infromation
```

```python
df['Order Date'] = pd.to_datetime(df['Order Date'])
df['Ship Date'] = pd.to_datetime(df['Ship Date'])
df.info()
```

```python
#removing outliers - too small or too large values to be removed 
df = df[(df['Sales']>=25) & (df['Sales']< 100000)]
df.info()
```

```python
#create separate tables from single dataset 
customer_df = df[['Customer ID','Customer Name','Segment','Country','City','State','Postal Code','Region']]
customer_df = customer_df.drop_duplicates(subset = "Customer ID")
customer_df.info()
                                          
```

```python
product_df = df[['Product ID','Category','Sub-Category','Product Name']]
product_df =product_df.drop_duplicates(subset ="Product ID")
product_df.info()
```

```python
orders_df = df[['Row ID','Order ID','Ship Date','Ship Mode','Customer ID','Product ID','Sales','Quantity','Discount','Profit']]   
orders_df.info()
```

```python
with pd.ExcelWriter("D:\\New_order\\Project1.xlsx",engine = 'xlsxwriter') as writer:
    customer_df.to_excel(writer,sheet_name = 'customer',index = False)
    product_df.to_excel(writer,sheet_name = 'product',index = False)
    orders_df.to_excel(writer,sheet_name = 'orders',index = False)
print("File created")
```
